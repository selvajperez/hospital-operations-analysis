### Limpieza y estandarización de nombres de columnas

Dado que el DataFrame se recarga, es importante aplicar nuevamente la limpieza de nombres de columnas para asegurar consistencia en el análisis.

```python
# Limpieza profesional de nombres de columnas
# Reaplicamos esta limpieza para el 'df' recién cargado.

df.columns = (
    df.columns
      .str.strip() # Elimina espacios en blanco al inicio y final
      .str.lower() # Convierte a minúsculas
      .str.replace(" ", "_", regex=False) # Reemplaza espacios con guiones bajos
      .str.replace("/", "_", regex=False) # Reemplaza barras con guiones bajos
      .str.replace("-", "_", regex=False) # Reemplaza guiones con guiones bajos
      .str.replace("__", "_", regex=False) # Reemplaza dobles guiones bajos con uno solo (por si acaso)
)

print("Nombres de columnas estandarizados:")
print(df.columns.tolist())
```

```python
# Pareto de Revenue

pareto = (
    df.groupby("case_type")["revenue"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

pareto["%"] = pareto["revenue"] / pareto["revenue"].sum() * 100
pareto["% Acumulado"] = pareto["%"].cumsum()

display(pareto)
```

```python
pareto = (
    df[["case_no","revenue"]]
      .sort_values("revenue", ascending=False)
      .reset_index(drop=True)
)

pareto["Revenue Acumulado"] = pareto["revenue"].cumsum()
pareto["% Acumulado"] = (
    pareto["Revenue Acumulado"] /
    pareto["revenue"].sum() * 100
)

display(pareto.head(20))
```

# Conclusiones del EDA

## Calidad de los datos

- 500 registros y 21 variables.
- No se detectaron registros duplicados.
- Se identificaron valores nulos únicamente en tres columnas (Discharge Time, Discharge Before 12PM e InsurancePlanName).
- Se normalizaron los nombres de las columnas para facilitar el análisis.

## Hallazgos principales

- El 93% del revenue proviene de pacientes Inpatient (IP).
- Existe una correlación fuerte (0,77) entre Length of Stay (LOS) y Revenue.
- La severidad presenta una relación muy débil con el revenue.
- PureWin Insurance concentra el 42,2% de los casos.
- El 4% de los casos genera aproximadamente el 26,5% del revenue total.

## Hipótesis validadas

- Existe concentración de ingresos en un único pagador (PureWin).
- La duración de la internación tiene mayor impacto en los ingresos que la severidad.

## Hipótesis descartadas o pendientes

- No fue posible confirmar que el estado "Inactive" de los médicos represente un riesgo legal o de facturación.
- Las diferencias observadas entre especialidades requieren información adicional del negocio para ser interpretadas correctamente.

## KPIs recomendados

- Revenue por Case Type.
- Revenue por Specialty.
- Revenue por Insurance Payer.
- Average LOS.
- CMI promedio por Specialty.
- % de altas antes de las 12 PM.
